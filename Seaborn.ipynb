# Seaborn NOTES
Seaborn is a Python library used for making charts and graphs easily.
Seaborn is designed to simplify the process of visualizing structured datasets by offering:

Built-in support for statistical aggregation

Integration with Pandas DataFrames

High-level functions for complex visualizations
