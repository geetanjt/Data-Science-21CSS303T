    # MULTIBAR CHART
   

```python
import matplotlib.pyplot as plt
w=0.4
x=['CSE','AIML','CC','CSBS','DS']
bar1=[0,1,2,3,4]
boys=[60,65,70,80,67]
girls=[55,59,60,70,50]
plt.bar(bar1,boys,color='r',width=w)
plt.bar(bar1,girls,color='g',width=w)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
w=0.4
x=['CSE','AIML','CC','CSBS','DS']
bar1=np.arange(len(x))
print(bar1)
bar2=[i+w for i in bar1]

print(bar2)
boys=[60,65,70,80,67]
girls=[55,59,60,70,50]
plt.bar(bar1,boys,color='r',width=w,label='boys')
plt.bar(bar2,girls,color='g',width=w,label='girls')
plt.xticks(bar1+w/2,x)
plt.xlabel('Branch',fontsize=20)
plt.ylabel(' no of students')
plt.title("STUDENTS DATA")
plt.legend()
plt.show()
```

```python
day=[1,2,3,4,5,6,7]
no=[2,3,1,4,5,3,6]
colors=[10,20,30,40,50,60,70]
sizes=[400,200,400,300,200,100,600]
plt.scatter(day,no,c=colors,cmap="viridis",s=sizes)
t=plt.colorbar()
t.set_label('color')
plt.show()
```
