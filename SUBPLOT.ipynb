```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[1,2,3,4]
plt.subplot(1,2,1)
plt.plot(x,y,color='r')
plt.pie([1],colors='g')
plt.subplot(1,2,1)
x2=[10,20,30,40]
plt.pie(x2)

x1=["x","y","c","d"]
y1=[1,2,3,4]
plt.bar(x1,y1)

plt.show()
```
